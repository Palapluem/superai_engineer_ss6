### duplicate โดยใช้ window_start_ts + window_end_ts + session_id เป็น key

````python
import json
from collections import defaultdict

# ── 1. Load fixed data ──────────────────────────────────────────────
with open("/content/windows.json") as f:
    content = f.read()

start = content.find('{')
end   = content.rfind(']')
data  = json.loads('[' + content[start:end+1])
print(f"✅ Loaded {len(data)} windows")

# ── 2. Find duplicates ──────────────────────────────────────────────
seen   = {}   # key → first index
dupes  = []   # list of duplicate indices

for i, w in enumerate(data):
    key = (
        w.get("window_start_ts", ""),
        w.get("window_end_ts",   ""),
        w.get("session_id",      ""),
    )
    if key in seen:
        dupes.append((i, seen[key], key))
    else:
        seen[key] = i

print(f"\n🔍 Found {len(dupes)} duplicates")
if dupes:
    print("\nSample duplicates (first 5):")
    for dup_idx, orig_idx, key in dupes[:5]:
        print(f"  idx {dup_idx} is duplicate of idx {orig_idx}")
        print(f"    start={key[0]}  session={key[2]}")
````

````python
# ── 3. Replace duplicates with {} ───────────────────────────────────
cleaned = []
dup_indices = {i for i, _, _ in dupes}

for i, w in enumerate(data):
    if i in dup_indices:
        cleaned.append({})          # replace with empty object
    else:
        cleaned.append(w)

# ── 4. Save ─────────────────────────────────────────────────────────
out_path = "/content/windows_deduped.json"
with open(out_path, "w", encoding="utf-8") as f:
    json.dump(cleaned, f, ensure_ascii=False, indent=2)

empty_count = sum(1 for w in cleaned if w == {})
print(f"\n✅ Saved to {out_path}")
print(f"   Total entries : {len(cleaned)}")
print(f"   Replaced ({{}}): {empty_count}")
print(f"   Valid windows : {len(cleaned) - empty_count}")
````

````python
import json

# ── Class mapping ────────────────────────────────────────────────────
CLASS_MAP = {
    "ล้มแบบค่อยๆทรุด":      {"class_en": "slow_collapse_fall",      "category": "fall"},
    "ค่อยๆล้ม":              {"class_en": "gradual_fall",            "category": "fall"},
    "คนแก้เดิน":             {"class_en": "corrected_walking",       "category": "activity"},
    "ลุกยืนสลับนั่ง":        {"class_en": "stand_sit_alternating",   "category": "activity"},
    "เดินกระเพก":            {"class_en": "limping_walk",            "category": "activity"},
    "เดินปกติ":              {"class_en": "normal_walk",             "category": "activity"},
    "คนแก่จับของระหว่างทาง": {"class_en": "elderly_pick_up_object",  "category": "activity"},
    "ยืน":                   {"class_en": "standing",                "category": "static_activity"},
    "นอน":                   {"class_en": "lying_down",              "category": "static_activity"},
}

# ── Load windows_deduped.json ─────────────────────────────────────────
with open("/content/windows_deduped.json", encoding="utf-8") as f:
    data = json.load(f)

# ── Apply mapping ─────────────────────────────────────────────────────
unmapped = set()
for w in data:
    if not w:  # skip {} empty
        continue
    cls = w.get("class", "")
    if cls in CLASS_MAP:
        w["class_en"] = CLASS_MAP[cls]["class_en"]
        w["category"] = CLASS_MAP[cls]["category"]
    else:
        unmapped.add(cls)

# ── Summary ───────────────────────────────────────────────────────────
from collections import Counter
cat_counts = Counter(w.get("category", "UNKNOWN") for w in data if w)
print("✅ Category distribution:")
for cat, count in sorted(cat_counts.items()):
    print(f"   {cat:20s}: {count}")

if unmapped:
    print(f"\n⚠️  Unmapped classes: {unmapped}")
else:
    print("\n✅ All classes mapped successfully!")

with open("/content/windows_mapped.json", "w", encoding="utf-8") as f:
    json.dump(data, f, ensure_ascii=False, indent=2)
print("\n💾 Saved: /content/windows_mapped.json")
````

````python
import numpy as np

def extract_window_features(window_df, fs=50):
    """
    window_df: DataFrame ของ 1 window มี columns ax,ay,az,gx,gy,gz,press,mic,p2p
    fs: sampling rate (Hz)
    """
    ax = window_df["ax"].values
    ay = window_df["ay"].values
    az = window_df["az"].values
    gx = window_df["gx"].values
    gy = window_df["gy"].values
    gz = window_df["gz"].values

    # ── SVM (Signal Vector Magnitude) ─────────────────────────────
    svm = np.sqrt(ax**2 + ay**2 + az**2)

    # ── Jerk (rate of change of acceleration) ─────────────────────
    jerk = np.sqrt(np.diff(ax)**2 + np.diff(ay)**2 + np.diff(az)**2) * fs

    # ── Angular velocity magnitude ─────────────────────────────────
    omega = np.sqrt(gx**2 + gy**2 + gz**2)

    # ── Tilt angle (theta from vertical) ──────────────────────────
    theta = np.degrees(np.arctan2(np.sqrt(ax**2 + ay**2), az))

    # ── KII (kinematic impact index) ──────────────────────────────
    KII = svm * omega

    features = {
        # SVM
        "svm_mean":        np.mean(svm),
        "svm_std":         np.std(svm),
        "svm_max":         np.max(svm),
        "svm_min":         np.min(svm),
        # Jerk
        "jerk_mean":       np.mean(jerk),
        "jerk_std":        np.std(jerk),
        "jerk_max":        np.max(jerk),
        "jerk_min":        np.min(jerk),
        "jerk_energy":     np.sum(jerk**2),
        "jerk_sparsity":   np.sum(jerk > np.mean(jerk) + 2*np.std(jerk)) / len(jerk),
        # KII
        "KII_mean":        np.mean(KII),
        "KII_std":         np.std(KII),
        "KII_max":         np.max(KII),
        "KII_min":         np.min(KII),
        # Omega
        "omega_mean":      np.mean(omega),
        "omega_std":       np.std(omega),
        "omega_max":       np.max(omega),
        "omega_min":       np.min(omega),
        # Theta
        "theta_mean":      np.mean(theta),
        "theta_std":       np.std(theta),
        "theta_max":       np.max(theta),
        "theta_min":       np.min(theta),
        "theta_range":     np.max(theta) - np.min(theta),
        # Events
        "free_fall_n":     int(np.sum(svm < 0.3)),          # near-zero g
        "impact_n":        int(np.sum(jerk > 15)),           # high jerk spike
        "high_rot_n":      int(np.sum(omega > 100)),         # high rotation
        # Angular impulse
        "angular_impulse": np.trapz(omega) / fs,
        # GSI (simplified)
        "GSI":             np.mean(svm) * np.mean(omega) / 1000,
        # Pressure
        "press_delta":     window_df["press"].iloc[-1] - window_df["press"].iloc[0]
                           if "press" in window_df.columns else 0,
        "press_slope":     np.polyfit(range(len(window_df)),
                           window_df["press"].values, 1)[0]
                           if "press" in window_df.columns else 0,
        # Mic
        "mic_p2p_max":     window_df["p2p"].max()  if "p2p"  in window_df.columns else 0,
        "mic_p2p_mean":    window_df["p2p"].mean() if "p2p"  in window_df.columns else 0,
        # SpO2/HR → 0 ถ้าไม่มี external sensor
        "hr_mean": 0, "hr_max": 0, "hr_delta": 0, "hr_spike": 0,
        "spo2_min": 0, "spo2_mean": 0,
        "rmssd": 0, "sdnn": 0, "hr_accel": 0,
        "osi": 0, "css_max": 0, "css_mean": 0, "fcri": 0,
    }
    return features
````

````python
# ── Test กับ joined_imu_windows.csv ────────────────────────────────
df = pd.read_csv("/content/joined_imu_windows.csv")

# เอา 1 window มาทดสอบ
win0 = df[df["win_id"] == 0].copy()
feats = extract_window_features(win0)

print("Extracted vs Original comparison:")
print(f"{'Feature':<20} {'Extracted':>12} {'Original':>12} {'Match':>6}")
print("-" * 55)
check_cols = ["svm_mean", "svm_std", "jerk_mean", "omega_mean", "theta_mean", "KII_mean"]
for col in check_cols:
    extracted = round(feats[col], 4)
    original  = round(win0[f"win_{col}"].iloc[0], 4)
    match     = "✅" if abs(extracted - original) < 0.1 else "⚠️"
    print(f"{col:<20} {extracted:>12} {original:>12} {match:>6}")
````

````python
import os, json
import numpy as np
import pandas as pd

# ── Feature extraction function ────────────────────────────────────
def extract_window_features(window_df, fs=50):
    ax = window_df["ax"].values.astype(float)
    ay = window_df["ay"].values.astype(float)
    az = window_df["az"].values.astype(float)
    gx = window_df["gx"].values.astype(float)
    gy = window_df["gy"].values.astype(float)
    gz = window_df["gz"].values.astype(float)

    svm   = np.sqrt(ax**2 + ay**2 + az**2)
    jerk  = np.sqrt(np.diff(ax)**2 + np.diff(ay)**2 + np.diff(az)**2) * fs
    omega = np.sqrt(gx**2 + gy**2 + gz**2)
    theta = np.degrees(np.arctan2(np.sqrt(ax**2 + ay**2), az))
    KII   = svm * omega

    jerk_safe = jerk if len(jerk) > 0 else np.array([0.0])

    return {
        "svm_mean":       np.mean(svm),   "svm_std":    np.std(svm),
        "svm_max":        np.max(svm),    "svm_min":    np.min(svm),
        "jerk_mean":      np.mean(jerk_safe), "jerk_std":  np.std(jerk_safe),
        "jerk_max":       np.max(jerk_safe),  "jerk_min":  np.min(jerk_safe),
        "jerk_energy":    float(np.sum(jerk_safe**2)),
        "jerk_sparsity":  float(np.sum(jerk_safe > np.mean(jerk_safe) + 2*np.std(jerk_safe)) / max(len(jerk_safe),1)),
        "KII_mean":       np.mean(KII),   "KII_std":    np.std(KII),
        "KII_max":        np.max(KII),    "KII_min":    np.min(KII),
        "omega_mean":     np.mean(omega), "omega_std":  np.std(omega),
        "omega_max":      np.max(omega),  "omega_min":  np.min(omega),
        "theta_mean":     np.mean(theta), "theta_std":  np.std(theta),
        "theta_max":      np.max(theta),  "theta_min":  np.min(theta),
        "theta_range":    float(np.max(theta) - np.min(theta)),
        "free_fall_n":    int(np.sum(svm < 0.3)),
        "impact_n":       int(np.sum(jerk_safe > 15)),
        "high_rot_n":     int(np.sum(omega > 100)),
        "angular_impulse":float(np.trapz(omega) / fs),
        "GSI":            float(np.mean(svm) * np.mean(omega) / 1000),
        "press_delta":    float(window_df["press"].iloc[-1] - window_df["press"].iloc[0]) if "press" in window_df.columns else 0.0,
        "press_slope":    float(np.polyfit(range(len(window_df)), window_df["press"].values.astype(float), 1)[0]) if "press" in window_df.columns else 0.0,
        "mic_p2p_max":    float(window_df["p2p"].max())  if "p2p" in window_df.columns else 0.0,
        "mic_p2p_mean":   float(window_df["p2p"].mean()) if "p2p" in window_df.columns else 0.0,
        # SpO2/HR — ไม่มี external sensor
        "hr_mean":0.0, "hr_max":0.0, "hr_delta":0.0, "hr_spike":0.0,
        "spo2_min":0.0, "spo2_mean":0.0, "rmssd":0.0, "sdnn":0.0,
        "hr_accel":0.0, "osi":0.0, "css_max":0.0, "css_mean":0.0, "fcri":0.0,
    }

# ── Sliding window config ──────────────────────────────────────────
WINDOW_SIZE = 40   # rows per window (same as windows.json)
STEP_SIZE   = 20   # 50% overlap

# ── Classes ที่ต้องการ extract ─────────────────────────────────────
TARGET_DIRS = {
    "fall":            ["gradual_fall"],
    "activity":        ["corrected_walking","elderly_pick_up_object",
                        "limping_walk","normal_walk","stand_sit_alternating"],
    "static_activity": ["standing","lying_down"],
}

LABEL_MAP = {
    "fall": 1,
    "activity": 0,
    "static_activity": 0,
}

all_windows = []

for category, classes in TARGET_DIRS.items():
    for class_en in classes:
        folder = f"/content/organized_data/{category}/{class_en}"
        if not os.path.exists(folder):
            print(f"⚠️  Not found: {folder}")
            continue

        csvs = [f for f in os.listdir(folder)
                if f.endswith(".csv") and "windows" not in f and "o2" not in f]

        for fname in sorted(csvs):
            path = os.path.join(folder, fname)
            df   = pd.read_csv(path)

            # ตรวจว่ามี IMU columns ครบ
            required = ["ax","ay","az","gx","gy","gz"]
            if not all(c in df.columns for c in required):
                print(f"⚠️  Skip (no IMU cols): {fname}")
                continue

            df = df.dropna(subset=required).reset_index(drop=True)

            # Sliding window
            n_windows = 0
            for start in range(0, len(df) - WINDOW_SIZE + 1, STEP_SIZE):
                chunk = df.iloc[start:start + WINDOW_SIZE].copy()
                feats = extract_window_features(chunk)

                feats["window_start_ts"] = chunk["pc_timestamp_iso"].iloc[0]  if "pc_timestamp_iso" in chunk.columns else ""
                feats["window_end_ts"]   = chunk["pc_timestamp_iso"].iloc[-1] if "pc_timestamp_iso" in chunk.columns else ""
                feats["session_id"]      = fname.replace(".csv","")
                feats["class"]           = class_en
                feats["class_en"]        = class_en
                feats["category"]        = category
                feats["label"]           = LABEL_MAP[category]
                feats["source"]          = "original"

                all_windows.append(feats)
                n_windows += 1

            print(f"  ✅ {class_en}/{fname}: {len(df)} rows → {n_windows} windows")

# ── Save ───────────────────────────────────────────────────────────
new_windows_df = pd.DataFrame(all_windows)
new_windows_df.to_csv("/content/windows_extracted.csv", index=False)

print(f"\n✅ windows_extracted.csv")
print(f"   Total windows : {len(new_windows_df)}")
print(f"\n📊 Windows per class:")
print(new_windows_df.groupby(["category","class_en"])["label"].count().to_string())
````

````python
# ── Merge กับ windows_mapped.json (fall windows เดิม) ──────────────
with open("/content/windows_mapped.json", encoding="utf-8") as f:
    existing = pd.DataFrame([w for w in json.load(f) if w])

# Rename columns ให้ตรงกัน
existing = existing.rename(columns={"class": "class_th"})
existing["class_en"] = existing["class_en"] if "class_en" in existing.columns else existing["class_th"]

combined = pd.concat([existing, new_windows_df], ignore_index=True)
combined.to_csv("/content/windows_all.csv", index=False)

print(f"✅ windows_all.csv (windows.json + extracted)")
print(f"   Total : {len(combined)} windows")
print(f"\n📊 Final distribution:")
print(combined.groupby(["category","class_en"])["label"].count().to_string())
````

````python
import os
import shutil

FOLDER_MAPPING = {
    # Thai name → (english_name, category)
    "ล้มแบบค่อยๆทรุด":         ("slow_collapse_fall",        "fall"),
    "ค่อยๆล้ม":                 ("gradual_fall",              "fall"),
    "คนแก้เดิน":                ("corrected_walking",         "activity"),
    "ลุกยืนสลับนั่ง":           ("stand_sit_alternating",     "activity"),
    "เดินกระเพก":               ("limping_walk",              "activity"),
    "เดินปกติ":                 ("normal_walk",               "activity"),
    "คนแก่จับของระหว่างทาง":    ("elderly_pick_up_object",    "activity"),
    "ยืน":                      ("standing",                  "static_activity"),
    "นอน":                      ("lying_down",                "static_activity"),
}

input_dir  = "/content/raw_data"
output_dir = "/content/organized_data"

moved, skipped = 0, 0

for thai_name, (eng_name, category) in FOLDER_MAPPING.items():
    src_folder = os.path.join(input_dir, thai_name)
    if not os.path.exists(src_folder):
        print(f"⚠️  Not found: {thai_name}")
        continue

    dst_folder = os.path.join(output_dir, category, eng_name)
    os.makedirs(dst_folder, exist_ok=True)

    # Walk recursively (handles subfolders like o2/)
    for root, dirs, files in os.walk(src_folder):
        rel_root = os.path.relpath(root, src_folder)  # e.g. "." or "o2"
        for filename in files:
            src_file = os.path.join(root, filename)

            # Rename: prepend eng_name to filename, keep subfolder prefix
            if rel_root == ".":
                new_name = f"{eng_name}_{filename}"
            else:
                new_name = f"{eng_name}_{rel_root}_{filename}"

            dst_file = os.path.join(dst_folder, new_name)
            shutil.copy2(src_file, dst_file)
            print(f" {thai_name}/{rel_root}/{filename} → {category}/{eng_name}/{new_name}")
            moved += 1

print(f"\n Done! {moved} files organized into {output_dir}")
````

````python
import pandas as pd

df = pd.read_csv("/content/windows_all.csv")

# Parse timestamps
df["window_start_ts"] = pd.to_datetime(df["window_start_ts"], utc=True, errors="coerce")
df["window_end_ts"]   = pd.to_datetime(df["window_end_ts"],   utc=True, errors="coerce")

# Duration per window (seconds)
df["duration_s"] = (df["window_end_ts"] - df["window_start_ts"]).dt.total_seconds()

# Total duration per class (sum all windows)
summary = df.groupby(["category", "class_en"]).agg(
    windows       = ("duration_s", "count"),
    total_min     = ("duration_s", lambda x: round(x.sum() / 60, 2)),
    avg_window_s  = ("duration_s", lambda x: round(x.mean(), 3)),
).reset_index()

summary = summary.sort_values(["category", "total_min"], ascending=[True, False])

print(f"{'Category':<18} {'Class':<30} {'Windows':>8} {'Total (min)':>12} {'Avg window (s)':>15}")
print("-" * 88)
for _, r in summary.iterrows():
    print(f"{r['category']:<18} {r['class_en']:<30} {r['windows']:>8} {r['total_min']:>12} {r['avg_window_s']:>15}")

print("-" * 88)
total_min = summary["total_min"].sum()
print(f"{'TOTAL':<50} {df.shape[0]:>8} {total_min:>12.2f} min")
````
