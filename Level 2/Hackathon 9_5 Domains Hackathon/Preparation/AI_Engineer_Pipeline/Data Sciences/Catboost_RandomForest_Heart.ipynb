```python
def clean_data(data):
    # Check if duplicated
    if data.duplicated().sum():
        data = data.drop_duplicated()
    data = data.dropna()
    df_yes = data[data['History of HeartDisease or Attack'] == 'Yes'].sample(n=len(data[data['History of HeartDisease or Attack'] == 'Yes']), random_state=42)
    df_no = data[data['History of HeartDisease or Attack'] == 'No'].sample(n=18000, random_state=42)
    data = pd.concat([df_yes, df_no], ignore_index=True)

    # Try fillna with median on BMI
    # data['Body Mass Index'] = data['Body Mass Index'].fillna(data['Body Mass Index'].median())

    # Remove Outliers
    numeric_cols = data.select_dtypes(include='number').columns
    for col in numeric_cols:
        Q1 = data[col].quantile(0.15)
        Q3 = data[col].quantile(0.85)
        IQR = Q3 - Q1
        print(Q1 - 1.5 * IQR)
        print(Q3 + 1.5 * IQR)
        outliers = data[(data[col] < Q1 - 1.5 * IQR) | (data[col] > Q3 + 1.5 * IQR)]
        print(f"{col}: {len(outliers)} potential outliers")

    # Input Outlier values
    data = data[data['Body Mass Index'] >= 5.914999999999996]
    data = data[data['Body Mass Index'] <= 50.115]
    return data
```

```python
def create_new_features(data):

    binary_map = {"Yes": 1, "No": 0}
    binary_cols = [
        "High Blood Pressure", "Told High Cholesterol", "Cholesterol Checked",
        "Smoked 100+ Cigarettes", "Diagnosed Stroke", "Diagnosed Diabetes",
        "Leisure Physical Activity", "Heavy Alcohol Consumption",
        "Health Care Coverage", "Doctor Visit Cost Barrier",
        "Difficulty Walking", "Vegetable or Fruit Intake (1+ per Day)"
    ]

    for col in binary_cols:
        # Create a new binary column with suffix '_bin'
        if data[col].dtype == object:
            data[col + "_bin"] = data[col].map(binary_map)
        else:
            data[col + "_bin"] = train_data[col]

    def categorize_bmi(bmi): # GOOD
        if bmi < 18.5:
            return "Underweight"
        elif bmi < 25:
            return "Normal weight"
        elif bmi < 30:
            return "Overweight"
        elif bmi < 40:
            return "Obese"
        else:
            return 'Morbidly Obese'

    data['BMI Category'] = data['Body Mass Index'].apply(categorize_bmi)
    category_order = ['Underweight', 'Normal weight', 'Overweight', 'Obese', 'Morbidly Obese']
    cat_type = pd.CategoricalDtype(categories=category_order, ordered=True)
    data['BMI Category'] = data['BMI Category'].astype(cat_type).cat.codes

    category_order = ['Never attended school', 'Elementary', 'Some high school', 'High school graduate', 'Some college or technical school','College graduate']
    cat_type = pd.CategoricalDtype(categories=category_order, ordered=True)
    data['education_ordinal'] = data['Education Level'].astype(cat_type).cat.codes

    category_order = ['Less than $10,000', '($10,000 to less than $15,000', '$15,000 to less than $20,000' ,
                      '$20,000 to less than $25,000', '$25,000 to less than $35,000' , '$35,000 to less than $50,000' ,
                      '$50,000 to less than $75,000', '$75,000 or more']
    cat_type = pd.CategoricalDtype(categories=category_order, ordered=True)
    data['income_ordinal'] = data['Income Level'].astype(cat_type).cat.codes

    category_order = ['Less than $10,000', '($10,000 to less than $15,000', '$15,000 to less than $20,000' ,
                      '$20,000 to less than $25,000', '$25,000 to less than $35,000' , '$35,000 to less than $50,000' ,
                      '$50,000 to less than $75,000', '$75,000 or more']
    cat_type = pd.CategoricalDtype(categories=category_order, ordered=True)
    data['income_ordinal'] = data['Income Level'].astype(cat_type).cat.codes

    category_order = ['Male', 'Female']
    cat_type = pd.CategoricalDtype(categories=category_order, ordered=True)
    data['Sex'] = data['Sex'].astype(cat_type).cat.codes

    data['Obesity Risk'] = ((data['Body Mass Index'] >= 30) & (data['Leisure Physical Activity'] == 'No')).astype(int)


    columns_to_drop = [
        "High Blood Pressure", "Told High Cholesterol", "Cholesterol Checked",
        "Smoked 100+ Cigarettes", "Diagnosed Stroke", "Diagnosed Diabetes",
        "Leisure Physical Activity", "Heavy Alcohol Consumption",
        "Health Care Coverage", "Doctor Visit Cost Barrier",
        "Difficulty Walking", "Vegetable or Fruit Intake (1+ per Day)",
        "Leisure Physical Activity","General Health","Income Level",'Education Level'
    ]
    data = data.drop(columns=columns_to_drop)
    return data
```

# Train MODEL

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from catboost import CatBoostClassifier
from sklearn.metrics import fbeta_score as f2

import warnings
warnings.filterwarnings('ignore')
```

```python
features = train_data.drop(['ID', 'History of HeartDisease or Attack'], axis=1)
target = train_data['History of HeartDisease or Attack']

X_train, X_val, Y_train, Y_val = train_test_split(
    features,
    target,
    random_state=3200,
    test_size=0.20)


X_train.shape, X_val.shape
```

```python
model = CatBoostClassifier(iterations=2,
                           learning_rate=1,
                           depth=2)
```

```python
scaler = StandardScaler()
scaler.fit(X_train)
X_train = scaler.transform(X_train)
X_val = scaler.transform(X_val)
```

```python
model = CatBoostClassifier(verbose=5,
                           iterations=1000,
                           loss_function='Logloss',
                           early_stopping_rounds=100,
                           depth = 8,
                           custom_metric=['AUC'])

model.fit(X_train, Y_train,
          eval_set=(X_val, Y_val))

y_train = model.predict(X_train)
y_val = model.predict(X_val)
```

```python
print("Training F2 : ", f2(Y_train, y_train,average=None, beta=0.5))
print("Validation F2: ", f2(Y_val, y_val,average=None, beta=0.5))
```

# WITH RANDOMFOREST

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report

# Step 1: Drop ID and separate features/target
X = train_data.drop(columns=['ID', 'History of HeartDisease or Attack'])
y = train_data['History of HeartDisease or Attack']

# Step 2: Ensure all features are numeric
X = X.select_dtypes(include=['number'])

# Step 3: Split the data
X_train, X_val, y_train, y_val = train_test_split(X, y, test_size=0.2, random_state=3200)

# Step 4: Scale features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_val_scaled = scaler.transform(X_val)

# Step 5: Train Random Forest
model = RandomForestClassifier(n_estimators=100, class_weight='balanced', random_state=42)
model.fit(X_train_scaled, y_train)

# Step 6: Evaluate the model
y_val_pred = model.predict(X_val_scaled)
print(classification_report(y_val, y_val_pred))

# Step 7: Prepare test data (same columns and order as training)
test_X = test_data.drop(columns=['ID'], errors='ignore')  # avoid crash if 'ID' missing
test_X = test_X[X.columns]  # ensure same column order
test_X_scaled = scaler.transform(test_X)

# Step 8: Predict on test data
test_data['History of HeartDisease or Attack'] = model.predict(test_X_scaled)
```
