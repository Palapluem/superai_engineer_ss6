# Prompt Template

We can use the prompt in 2 ways in Langchain


1. Using langchain.prompts.ChatPromptTemplate with prompt | llm -> invoke

2. Using the langchain_core.messages, using llm(messages)

## LCEL Format : prompts.ChatPromptTemplate -> prompt_template | model_llm

````python
# https://python.langchain.com/api_reference/core/prompts/langchain_core.prompts.chat.ChatPromptTemplate.html#chatprompttemplate
prompt_template = prompts.ChatPromptTemplate([
    (
        "system", "You are a helpful assistant."
    ),
    (
        "human", "{text}"
    ),
])

# LangChain Expression Language (LCEL)
chain = prompt_template | model_llm
chain.invoke({"text": "Hello"})
````

````python
# https://python.langchain.com/docs/integrations/vectorstores/pinecone/
from pinecone import ServerlessSpec
import time

# In Vector DB, Index means Tables
index_name = 'tables-us'
existing_indexes = [
    index_info["name"] for index_info in pc.list_indexes()
]


# check if index already exists (it shouldn't if this is first time)
if index_name not in existing_indexes:

    # if does not exist, create index
    pc.create_index(
        name=index_name,

        # dimensionality of HF embedding
        dimension=768,

        metric='dotproduct',
        spec=ServerlessSpec(cloud="aws", region="us-east-1")
    )

    # wait for index to be initialized
    while not pc.describe_index(index_name).status['ready']:
        time.sleep(1)


# Target the index
dense_index  = pc.Index(index_name)
time.sleep(1)

# View stats for the index
dense_index.describe_index_stats()
````

````python
from tqdm.auto import tqdm
import numpy as np
import re
import unicodedata


# Helper function to convert numpy types to Python native types and handle nulls
def convert_to_native_types(obj):
    if obj is None:
        return ""  # Convert None to empty string

    elif isinstance(obj, np.ndarray):
        return obj.tolist()

    elif isinstance(obj, np.integer):
        return int(obj)

    elif isinstance(obj, np.floating):
        return float(obj)

    elif isinstance(obj, (list, tuple)):
        return [convert_to_native_types(item) for item in obj]

    elif isinstance(obj, dict):
        return {k: convert_to_native_types(v) for k, v in obj.items()}

    else:
        return obj
````

````python
# Function to create ASCII-compatible IDs
def create_safe_id(destination, name, idx):
    # Normalize and remove accents
    destination = unicodedata.normalize('NFKD', str(destination)).encode('ASCII', 'ignore').decode('ASCII')
    name = unicodedata.normalize('NFKD', str(name)).encode('ASCII', 'ignore').decode('ASCII')

    # Replace any remaining non-alphanumeric characters with underscores
    destination = re.sub(r'[^a-zA-Z0-9]', '_', destination)
    name = re.sub(r'[^a-zA-Z0-9]', '_', name)

    # Create ID with sanitized values
    return f"{destination}_{name}_{idx}"
````

````python
BATCH_SIZE = 100

# Store valid records
valid_ids = []
valid_embeds = []
valid_metadata = []

for i in tqdm(range(0, len(dataset_pandas), BATCH_SIZE)):
    i_end = min(len(dataset_pandas), i + BATCH_SIZE)
    # get batch of data
    batch = dataset_pandas.iloc[i:i_end]


    for j, (idx, row) in enumerate(batch.iterrows()):

        try:
            # Check if essential fields exist and are not None
            if (row['DESCRIPTION'] is None or 'description' not in row['DESCRIPTION'] or row['DESCRIPTION']['description'] is None):
                continue

            # Generate ASCII-safe ID
            record_id = create_safe_id(row['DESTINATION'], row['NAME'], idx)

            # Create text for embedding
            text_to_embed = f"Name: {row['NAME']}\nLocation: {row['DESTINATION']}\nDescription: {row['DESCRIPTION']['description']}"

            # Get embedding
            embed = huggingface_embedding.embed_documents([text_to_embed])[0]

            # Prepare metadata with safe defaults
            metadata_entry = convert_to_native_types({
                'text': row['DESCRIPTION'].get('description', ""),
                'name': row['NAME'] or "",
                'location': row['DESTINATION'] or "",
                'rating': row['RATING'] if row['RATING'] is not None else 0.0,
                'categories': row['DESCRIPTION'].get('categories', []) if row['DESCRIPTION'] else [],
                'reviews': row['REVIEWS'] if row['REVIEWS'] is not None else [],
                'url': row['URL'] or ""
            })

            # Add to valid records
            valid_ids.append(record_id)
            valid_embeds.append(embed)
            valid_metadata.append(metadata_entry)

        except Exception as e:
            print(f"Error processing row {idx}: {e}")
            continue


    # If we have valid records in this batch, upsert them
    if valid_ids:
        # Upsert in smaller sub-batches to avoid overflow issues
        SUB_BATCH_SIZE = 50
        for k in range(0, len(valid_ids), SUB_BATCH_SIZE):
            k_end = min(len(valid_ids), k + SUB_BATCH_SIZE)

            try:
                dense_index.upsert(
                    vectors=zip(
                        valid_ids[k:k_end],
                        valid_embeds[k:k_end],
                        valid_metadata[k:k_end]
                    )
                )
                print(f"Successfully upserted {k_end - k} records")

            except Exception as e:
                print(f"Error upserting batch {k}-{k_end}: {e}")
                # Debug the first problematic ID
                if len(valid_ids[k:k_end]) > 0:
                    print(f"First ID in batch: {valid_ids[k]}")

        # Clear lists for next batch
        valid_ids = []
        valid_embeds = []
        valid_metadata = []
````

````python
from langchain_pinecone import PineconeVectorStore

vectorstore = PineconeVectorStore(
  index = dense_index,
  embedding = huggingface_embedding,
  text_key = "text"
)
````

````python
query = "What is Vizcaya Museum and Gardens"

vectorstore.similarity_search(query=query, k=1)
````

````python
def augment_prompt(query: str) -> str:
  results = vectorstore.similarity_search(query=query, k=3)
  source_knowledge = "\n".join([x.page_content for x in results])
  augmented_prompt = f"""
  Using the contexts below, answer the query.

  Contexts:
  {source_knowledge}

  Query: {query}
  """

  return augmented_prompt
````
